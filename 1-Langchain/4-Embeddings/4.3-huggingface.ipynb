Sentence Transformers on Hugging Face

Hugging Face sentece-transformers is a Python framework for state-of-the-art sentence, text and image embedding. One of the embedding models in the HuggingFaceEmbeddings class. We have also added an alias for SentenceTransformerEmbeddings for users who are more familiar with directly using that package.

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2", model_kwargs={"device": "cpu"})
```

```python
text = "This is a test document."
query_result = embedding.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
doc_result = embedding.embed_documents([text,"This is not a test document."])
doc_result[0]
doc_result[1]
```
