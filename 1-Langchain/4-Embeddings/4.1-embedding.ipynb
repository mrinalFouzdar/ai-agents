Converting text into vectors

```python
import os
from dotenv import load_dotenv
load_dotenv() #load environment variables from .env file
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embedding = OpenAIEmbeddings(model="text-embedding-3-large")
embedding
```

```python
text ="This is a test document. It is used to test the embedding model."
query_result = embedding.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
embedding_1024=OpenAIEmbeddings(model="text-embedding-3-large", dimensions=1024)
```

```python
embedding_1024
```

```python
text ="This is a test document. It is used to test the embedding model."
query_result = embedding_1024.embed_query(text)
len(query_result)
```

```python
## Text Loader

from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_documents = text_splitter.split_documents(docs)
final_documents
```

```python
## Vector Embedding And Vector StoreDB
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_documents, embedding_1024)
db
```

```python
### Retrieve the results from query vectorstore db
query ="all the easier for us to conduct ourselves as belligerents"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```
